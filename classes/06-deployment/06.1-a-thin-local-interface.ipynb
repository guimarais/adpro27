## From a module to something another person can use

A **UI** is the interface a person sees and operates. Here a browser displays a page, while a Python **server** runs Streamlit and our application code. An **entrypoint** is the script the server starts: `streamlit_app.py`. Importing that file here reads constants and function definitions; it does not start a server.

The flow is small: browser choice → Streamlit script → supplied data functions → displayed result. `Observation` validates records. `load_observations()` loads the local snapshot. `summarize()` calculates count/total/mean. The UI chooses which records are visible and formats the answer. It does not copy the data calculation or teach a second analysis implementation.

Open the script in VS Code alongside the notebook. Identify `st.title`, `st.selectbox`, `st.metric`, and the calls to the existing functions. Streamlit reruns the script when a widget changes; variables are created again. Session state retains only the small example result and live-use flag for that browser session. A server restart/new session does not preserve those values.

A browser-based app is different from a browser notebook server. Continue learning in VS Code's notebook editor; run this application's server from its terminal.

## Configuration belongs to the server

**Configuration** selects runtime behavior without changing source history. A **secret** is a sensitive value such as an API credential. In this app, missing `[ai]` configuration means live AI stays disabled; data and saved replay continue to work. A missing optional key should not make a public saved-data service unusable.

Spend the first **10 of the 30 minutes** reading the app's supplied configuration branch and [deployment guide section 5](DEPLOYMENT.md). Copy the tracked disabled example to the ignored `.streamlit/secrets.toml` only if practicing local configuration; leave `enabled = false`. Confirm the real path is ignored. Required completion never needs actual credentials. The example is safe to track because it contains no real secret; the real file must stay local or in hosting's Secrets settings.

An explicitly enabled but incomplete setup produces a short warning while preserving data/replay. The app never prints keys, passcodes, or raw API exception bodies. In a hosted app, the browser is not an appropriate place to store an instructor's key. A remote localhost also does not refer to your laptop.

Optional live demonstrations require private hosting access, server settings, an instructor access code, and a separately verified model. The session flag permits only one live run in that browser session; it is not a global spending cap. A new session or server can run again. The harness still limits actual tools and total model requests. Read those limits before enabling optional live use; no paid access is required by this lesson.

## Logs should point to an action

Spend the next **10 minutes**, including Exercise 06.5.H1 below, inspecting the supplied logger calls and one actual local/hosted log if available. A useful line identifies the event and useful non-sensitive context: `view release=06.0 rows=2` tells you which release/filter result is running. `data_load_failed kind=ValueError` points to data validation. `live_ai_unavailable reason=incomplete_configuration` points to optional settings.

A log line is evidence, not a diagnosis by itself. State the stage, cause hypothesis, smallest fix, and next check. The UI's data-error message asks you to restore fixtures; it does not expose Python traceback internals to every visitor. Keep the diagnostic event useful without copying exception text that might contain private provider headers/bodies or keys. A redacted small log excerpt is better evidence than a screenshot of an entire secret-filled settings screen.

The saved invalid-arguments trace above is a harness failure, not a hosting failure. Distinguish application policy/schema stops from dependency/startup problems and from provider errors.

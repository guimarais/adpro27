### Complete baseline

This Report stores validated Observation objects. It deliberately supplies only collection and total behavior. You can copy/extend these definitions in your answer. A blank answer does not break the demonstration or prevent another task from being attempted.

```python
from pydantic import BaseModel, Field, ValidationError


class Observation(BaseModel):
    """A nonempty label and a finite, nonnegative quantity."""
    label: str = Field(min_length=1)
    value: float = Field(ge=0, allow_inf_nan=False)


class Report:
    """Collect observations already validated at the input boundary."""

    def __init__(self, name):
        self.name = name
        self.records = []

    def add(self, observation):
        self.records.append(observation)

    def total(self):
        return sum(observation.value for observation in self.records)
```

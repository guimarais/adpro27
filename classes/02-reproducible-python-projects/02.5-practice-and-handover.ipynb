The next cell executes these two supplied local programs with the selected Python and compares their parsed JSON. It does not install anything, modify source, or call an external service.

```python
import json
import subprocess

original = subprocess.run([sys.executable, "practice/monolithic_report.py"], capture_output=True, text=True, check=True)
modular = subprocess.run([sys.executable, "-m", "report_app"], capture_output=True, text=True, check=True)
original_data = json.loads(original.stdout)
modular_data = json.loads(modular.stdout)
print("Original:", original_data)
print("Modular:", modular_data)
assert original_data == modular_data == {"name": "class-02", "count": 2, "total": 6, "mean": 3}
```

### Readability explains the contract

A useful name tells the reader what a value or operation represents. A small function has a clear job and explicit inputs. Type hints document intended kinds of values, while a docstring explains meaning, assumptions, and important edge behavior. Hints alone still do not validate arbitrary runtime input; keep the Observation boundary from the example.

The following short function takes validated objects and returns labels in the same order. Its docstring explains behavior rather than restating every line.

````python
from report_app.models import Observation


def record_labels(records: list[Observation]) -> list[str]:
    """Return labels in input order without modifying the observations."""
    return [record.label for record in records]


records = [Observation(label="north", value=2.5), Observation(label="south", value=3.5)]
print(record_labels(records))
assert record_labels(records) == ["north", "south"]
````

### Ruff: two different operations

Ruff provides lint feedback and formatting. In the class terminal:

```sh
uv run --locked ruff check report_app solutions tests
uv run --locked ruff format --check report_app solutions tests
```

The first checks the selected rules, such as undefined or unused names. The second reports whether formatting differs from the formatter's output. `ruff format file.py` writes formatting changes, so use it on your own scratch copy. The small configuration in `pyproject.toml` supplies the Python target and rule selection.

Passing these checks does not establish that the calculation is correct or its names are meaningful. Compare behavior for ordinary and empty input, and read the interface. Avoid a stack of competing formatters and scoring tools.

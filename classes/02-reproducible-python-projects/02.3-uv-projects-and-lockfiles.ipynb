## In class · 60–80 minutes — Three files/directories, three roles

| Item | Role |
| --- | --- |
| `pyproject.toml` | Project metadata and declared direct dependency requirements |
| `uv.lock` | The resolved dependency information managed by uv |
| `.venv/` | The generated local environment used to execute code |

A declared range such as `pydantic>=2,<3` allows compatible versions. The lock records an exact resolution for reproducible installation. Do not edit the lockfile manually to pretend the resolver chose something else.

````python
import tomllib

manifest = tomllib.loads(Path("pyproject.toml").read_text())
lock = tomllib.loads(Path("uv.lock").read_text())
print("Project:", manifest["project"]["name"])
print("Direct runtime requirements:", manifest["project"]["dependencies"])
print("Development tools:", manifest["dependency-groups"]["dev"])
resolved = {package["name"]: package["version"] for package in lock["package"] if "version" in package}
print("Resolved Pydantic:", resolved["pydantic"])
````

### The small everyday workflow

- `uv add package` changes project dependency metadata and updates the resolution/environment.
- `uv run --locked ...` runs in the project environment while requiring the lock to remain compatible.
- `uv sync --locked` creates/synchronizes the environment from the existing project and lock.
- `uv add --dev ruff` declares a development tool, rather than a runtime import used by the application.

`uv pip install` is a pip-style installation interface. It is not interchangeable with declaring a project dependency through `uv add`. If someone can run your code only because they manually installed an undeclared package, the handover is incomplete.

### Start a disposable project with no Git requirement

Run these commands manually in a new scratch location, not over the class folder:

```sh
uv init --bare --vcs none dependency-demo
cd dependency-demo
uv add pydantic
uv run --locked python -c "import pydantic; print(pydantic.__version__)"
```

`--bare` keeps the scaffold small; `--vcs none` avoids introducing version-control setup before class 03. Pydantic is already used in the class environment, but this new project starts without that dependency, so you can see what declaring it changes.

Package resolution/installation can require network access. The notebook cells themselves only inspect the supplied files. The exact chosen version can change when creating a new lock later; the point is to inspect the declaration and the resolution, not memorize a version.

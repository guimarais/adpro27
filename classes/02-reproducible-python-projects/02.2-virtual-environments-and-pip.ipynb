## In class · 30–50 minutes — Which Python is running?

An installed package is not simply available to every Python on a machine. Different interpreters and virtual environments can have different packages. A **virtual environment** gives a project its own interpreter context and installation location, using a base Python installation.

`sys.executable` identifies the actual executable. `sys.prefix` identifies its environment; `sys.base_prefix` identifies the base installation. A familiar version number alone does not show that you chose the intended environment.

````python
import importlib.metadata
import importlib.util
import pydantic

print("Executable:", sys.executable)
print("Environment:", sys.prefix)
print("Base:", sys.base_prefix)
print("Pydantic version:", importlib.metadata.version("pydantic"))
print("Pydantic location:", pydantic.__file__)
print("pip module present:", importlib.util.find_spec("pip") is not None)
assert sys.prefix != sys.base_prefix
````

### Create a scratch environment manually

In a terminal, create a new `pip-demo` folder, enter it, and run `python -m venv .venv`. Use a Python installation with venv support. Do not run this over the class's supplied `.venv`.

On Linux/macOS, from that new folder:

```sh
python -m venv .venv
.venv/bin/python -m pip --version
.venv/bin/python -m pip install pydantic
.venv/bin/python -c "import sys, pydantic; print(sys.executable); print(pydantic.__file__)"
```

On Windows PowerShell, from that new folder:

```powershell
python -m venv .venv
.\.venv\Scripts\python.exe -m pip --version
.\.venv\Scripts\python.exe -m pip install pydantic
.\.venv\Scripts\python.exe -c "import sys, pydantic; print(sys.executable); print(pydantic.__file__)"
```

Installation can require network access. These commands use the environment's executable directly, so you do not need to change PowerShell execution policy to activate it. If venv creation is unavailable in your Python installation, ask teaching support rather than installing into system Python.

### Why `python -m pip`?

It runs pip through the named Python. A bare `pip` command can resolve to another installation on your PATH. Use the exact executable when diagnosing a mismatch. This is an environment issue, not a reason to copy package folders manually.

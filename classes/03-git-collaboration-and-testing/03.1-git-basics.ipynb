## In class · 0–20 minutes — Start with Git

**Git** records versions of a project as commits in a local repository. **GitHub** hosts repositories and adds collaboration tools. You can inspect and commit locally without a GitHub account or internet connection. A commit records a snapshot and metadata; it is not a submission to a remote server.

Three states matter immediately:

| Place | Meaning | Inspect it |
| --- | --- | --- |
| Working files | What you are editing now | `git status`, `git diff` for tracked unstaged changes |
| Staging index | The proposed contents of the next commit | `git diff --staged` |
| Local history | Snapshots already committed | `git log --oneline` |

`git add filename` copies that file's current contents into the index. If you edit it again, the newer edits are not automatically staged. `git commit` records the index, so inspect the staged diff before committing. A file can have both staged and unstaged changes.

## Guided follow-up — Independent work · 15 minutes

In your scratch repository, stage a README edit and then add another line without staging it. Compare both diffs. Predict which line would be committed, then inspect the staged content before deciding what to include. Make a deliberate coherent commit or leave the second edit as explained working state.

Use `git log --oneline -3` and `git show HEAD` to connect a commit message to its actual diff. A green test or a familiar filename does not prove that the intended version was staged. Git records what you select; the next parts add review and behavioral evidence.

Beginner recovery for a committed ordinary change: first inspect the relevant commit with `git show <commit-id>`, then use `git revert <commit-id>` in the scratch repository. Revert creates another commit that undoes that commit's changes. It can conflict; inspect the result and test it. Practice `git revert --no-edit HEAD` only after making a disposable, ordinary single-parent commit with a clear change. Do not use a merge commit for this introductory drill.

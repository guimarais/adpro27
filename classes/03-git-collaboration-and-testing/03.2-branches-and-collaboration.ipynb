## In class · 30–50 minutes — A branch names a line of work

A **branch** is a movable name for a commit. New commits on the current branch move that name forward. It is not a separate copy of the whole project on disk. `git switch -c feature/summary-docs` creates and switches to a branch; `git switch main` returns to main. Commit or otherwise deliberately handle your edits before switching; a dirty working tree can make switching confusing or prevent it.

In a separate scratch repository containing the supplied Git starter and an initial commit:

```sh
git switch -c feature/summary-docs
# Edit README.md to explain None versus a measured zero.
git diff
git add README.md
git diff --staged
git commit -m "Clarify missing data and zero measurements"
git log --oneline --graph --all
```

Read the graph as commit ancestry. Two branches can point to the same commit until one receives new work. A merge integrates histories. If the target has not diverged, integration can be a fast-forward; divergent histories may require a merge commit and may produce a conflict. Do not infer a conflict simply from two people editing the same file.

Optional illustration: [branch/review process image and attribution](IMAGE_SUGGESTIONS.md). The suggested GitLab diagram uses “merge request”; GitHub calls the review object a “pull request”.

### Remote does not mean synchronized automatically

A **remote** such as `origin` names another repository's location. `origin/main` is your locally stored remote-tracking reference, updated when you fetch. It is not a live view of the server.

| Command | Effect |
| --- | --- |
| `git fetch origin` | Downloads commits and updates remote-tracking references; does not merge them into your current branch |
| `git push -u origin feature/summary-docs` | Sends that branch to a configured remote and records its upstream if accepted |
| `git pull --ff-only` | Fetches, then attempts fast-forward integration into the current branch; stops if histories have diverged |

A pull is not a promise to overwrite your files with the server's version. If fast-forward integration fails, inspect status and history and decide how to integrate with teaching support. Avoid force pushing as a beginner response to a rejected update.

For the online route, use a repository you own or an instructor-provided classroom repository and its actual URL. If no remote is configured, add that authorized URL manually with `git remote add origin <your-repository-url>`. Authentication depends on the hosting setup; never paste a token into a notebook or commit. Check the remote with `git remote -v` before pushing.

For a new Git-starter remote, create an empty GitHub repository without a generated README/license and publish the baseline with `git push -u origin main` before pushing the feature branch. The remote target must exist to open a PR against it. For an existing classroom repository, clone its agreed baseline and branch from that history; do not attach an unrelated scratch history to it. The review guide gives both routes.

### Pull requests turn a diff into a conversation

A pull request proposes changes from a source/compare branch into a target/base branch. On GitHub, check that the base is the intended `main` and the compare branch is your feature branch. Read the diff, explain the problem and resulting behavior, list checks actually run, and ask for specific review. A local commit or `git push` alone does not create that review conversation.

Use [the review guide](REVIEW_GUIDE.md) for the full manual online route and offline alternative. An offline packet contains the diff, context, expected behavior, test evidence, and a review question. It practices the reasoning and gives provisional completion. Use it when account/network access prevents the online exercise, then complete the actual hosted PR and peer review in the reserved 15 minutes of 03.5.H1. If access is still blocked, the instructor schedules those same reserved minutes later; hosted completion remains pending.

A useful comment names a behavior, its impact, and a concrete request: “This changes the empty mean from None to zero, so readers cannot distinguish missing data from a measured zero. Keep None and add an empty-input test.” That is easier to act on than “this looks wrong”. Small diffs make those claims easier to check.

## Independent work · 30-minute diagnosis and conflict task

Two kinds of disagreement need different evidence: an assertion reports behavior that differs from a requirement; a Git conflict reports content that Git could not combine automatically. Resolving one does not necessarily resolve the other.

The recorded conflict below supplies all the state needed for an offline reasoning exercise. It is plain text, not an active merge in your checkout. The marker names show two competing versions; neither marker is a decision about correctness.

```python
conflict = Path("practice/conflicts/README-conflicted.txt").read_text()
print(conflict)
assert "<<<<<<< HEAD" in conflict and ">>>>>>>" in conflict
```

Keep the explanation that matches the contract: an empty collection has mean None because no measurements exist. Remove the conflict markers and the rejected alternative; preserve useful surrounding text. In an actual scratch merge, Git considers the path resolved after you stage it, and the merge is completed by a commit. That only records your decision. Inspect and run checks to establish whether the decision was correct.

For a real manual conflict drill, use a separate scratch Git starter with a clean initial commit:

1. Make README.md contain exactly two lines: `Quantity report` and `Empty behavior is to be decided.` Commit that preparation on main.
2. Create `feature/empty-policy`; change its second line to `An empty collection has mean 0.` and commit.
3. Switch to main; change that same second line to `An empty collection has mean None because no measurements exist.` and commit.
4. From main, run `git merge feature/empty-policy`. Expect a conflict because both branches changed the same baseline line differently.
5. Inspect `git status`, edit README.md according to the contract, and inspect `git diff`. Run `git add README.md`, inspect `git diff --staged`, then `git commit -m "Resolve empty-mean contract conflict"`. Check status and the final content. The tiny starter's `summary.mean([])` must still be None.

This drill uses no remote, forced update, or history rewriting. If a different baseline produces no conflict, inspect the recorded example and explain the competing decisions rather than changing your actual course checkout.

## Data contracts — In class

`Observation` has a nonempty label and a nonnegative numeric value. `model_validate()` checks a raw dictionary; type hints alone do not do this. Pydantic's default mode can convert a numeric string, so validation policy is a deliberate choice.

The schema describes our application's measurement contract. Generating JSON Schema describes that structure for another system; it does not serialize a particular observation or establish factual correctness.

```python
from pydantic import ValidationError

observation = Observation.model_validate({"label": "south", "value": "3.5"})
print(observation.model_dump())
print("JSON Schema fields:", list(Observation.model_json_schema()["properties"]))

try:
    Observation.model_validate({"label": "west", "value": -2})
except ValidationError as error:
    print("Rejected input at field:", error.errors()[0]["loc"][0])
```
